## Task 1: Data Exploration and Preprocessing

## 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

###  2: Load Dataset

```python
df=pd.read_csv("/Users/siddharthkushwaha/Desktop/Cognifyz/Level_1/Dataset .csv")
```

### 3: Display First 5 Rows

```python
df.head()
```

### 4: Check Number of Rows and Columns

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

### Step 5: Check Column Names

```python
print("Column Names:")
print(df.columns.tolist())
```

###  6: Check Data Types

```python
print(df.dtypes)
```

### 7: Check Missing Values

```python
missing_values = df.isnull().sum()
print(missing_values)
```

### Step 8: Handle Missing Values

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

### Again check

```python
print(df.isnull().sum())
```

### Step 9: Check Data Types Again

```python
print(df.dtypes)
```

### 10: Analyze Aggregate Rating Distribution

```python
print(df["Aggregate rating"].describe())
```

### Step 11: Check Rating Frequency

```python
rating_counts = df["Aggregate rating"].value_counts().sort_index()

print(rating_counts)
```

### Step 12: Visualize Rating Distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["Aggregate rating"],bins=10,kde=True)

plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.show()
```

### Step 13: Check Class Imbalance

```python
rating_percentage = (df["Aggregate rating"].value_counts(normalize=True)
    .sort_index() * 100)

print("Rating distribution percentage:")
print(rating_percentage)
```

### Visualization:

```python
plt.figure(figsize=(10, 6))

sns.countplot(x=df["Aggregate rating"])

plt.title("Frequency of Aggregate Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.show()
```

### Step 14: Final Dataset Check

```python
print("Final dataset shape:", df.shape)

print("\nRemaining missing values:")
print(df.isnull().sum())

print("\nData types:")
print(df.dtypes)
```
