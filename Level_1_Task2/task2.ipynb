# Level 1 — Task 2: Descriptive Analysis

### Step 1 — Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2 — Load Cleaned Dataset

```python
df=pd.read_csv("/Users/siddharthkushwaha/Desktop/Cognifyz/Level_1_Task2/cleaned_zomato_dataset.csv")

df.head()
```

### Step 3 — Numerical Columns Identify Karo

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(numeric_columns)
```

### Step 4 — Basic Statistical Measures

```python
df[numeric_columns].describe()
```

### mean

```python
print("Mean:")
print(df[numeric_columns].mean())
```

Median:

```python
print("Median:")
print(df[numeric_columns].median())
```

Standard deviation

```python
print("Standard Deviation:")
print(df[numeric_columns].std())
```

### Step 5 — Country Code Distribution

```python
country_counts = df["Country Code"].value_counts()

print(country_counts)
```

#### Visualization:

```python
plt.figure(figsize=(10, 6))

sns.barplot(x=country_counts.index,y=country_counts.values)

plt.title("Number of Restaurants by Country Code")
plt.xlabel("Country Code")
plt.ylabel("Number of Restaurants")

plt.show()
```

### Step 6 — City Distribution

```python
city_counts = df["City"].value_counts()

print(city_counts.head(10))
```

### Top 10 cities visualize:

```python
top_cities = city_counts.head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_cities.values,y=top_cities.index)

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("City")

plt.show()
```

### Step 7 — Cuisines Distribution

```python
cuisine_counts = df["Cuisines"].value_counts()

print(cuisine_counts.head(10))
```

Top 10 cuisines:

```python
top_cuisines = cuisine_counts.head(10)

print("Top 10 Cuisines:")
print(top_cuisines)
```

### Step 8 — Top Cuisines Visualization

```python
plt.figure(figsize=(10, 6))

sns.barplot(x=top_cuisines.values,y=top_cuisines.index)

plt.title("Top 10 Cuisines by NUmber of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisines")

plt.show()
```

### Step 9 — Check Unique Values

Country:

```python
print("Total Restaurants:", len(df))

print("Unique Countries:", df["Country Code"].nunique())

print("Unique Cities:", df["City"].nunique())

print("Unique Cuisine Entries:", df["Cuisines"].nunique())

print("\nTop 5 Cities:")
print(city_counts.head(5))

print("\nTop 5 Cuisines:")
print(cuisine_counts.head(5))
```
